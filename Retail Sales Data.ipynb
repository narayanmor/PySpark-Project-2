```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Read the SuperStore CSV file from the specified volume path
# header=true uses the first row as column names
SuperStoredf=spark.read.format("csv").option("header","true").load("/Volumes/pyspark_project/spark/datafiles/SuperStoredata/superstore.csv")

# Register the DataFrame as a temporary view so it can be queried with SQL in subsequent cells
SuperStoredf.createOrReplaceTempView("SuperStoredf")

# Display the loaded DataFrame to preview its contents
display(SuperStoredf)
```

```python
%sql

select count(distinct customer_id) as Total_Customers from SuperStoredf 
```

```python
%sql
select count(distinct customer_id) as Total_Customers from SuperStoredf  where order_date between '2024-01-01' and '2024-01-31'
```

```python
dbutils.widgets.dropdown("time_period",'weekly',['weekly','monthly'])
# DBTITLE 1,Querying the DataFrame with SQL
# Query the DataFrame using SQL
```

```python
from datetime import datetime, timedelta, date
from pyspark.sql.functions import *

time_period = dbutils.widgets.get("time_period")
print(time_period)
today = date.today()
if time_period == 'weekly':
    start_date = today - timedelta(days=today.weekday(), weeks=1) - timedelta(days=1)
    end_date = start_date + timedelta(days=6)
else:
    first_day = today.replace(day=1)
    end_date = first_day - timedelta(days=1)
    start_date = first_day - timedelta(days=end_date.day)
    print(start_date,end_date)
```

```python
display(spark.sql(f"""select count(distinct customer_id) as Total_Customers from SuperStoredf  where order_date between 
'{start_date}' and '{end_date}'"""))
```
